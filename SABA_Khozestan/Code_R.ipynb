```r
library(package = dplyr)
library(package = ggplot2)

options(repr.plot.width = 10, repr.plot.height = 10, repr.plot.res = 300)
```

```r
data = read.csv(file = "Output/result.csv", header = TRUE)
head(x = data, n = 10)
```

```r
line1_1 = data_frame(x = c(0, 600), y = c(0, 600))

(ggplot(data = data) + 
 geom_point(mapping = aes(x = Precip, y = TerraClimate), size = 1) +
 geom_point(mapping = aes(x = Precip, y = ERA5_Monthly_Aggregate), color = 'red', size = 1) +
 facet_wrap('~Station') +
 geom_line(data = line1_1, mapping = aes(x = x, y = y), color = 'blue'))
```

```r
data_yearly = read.csv(file = "Output/data_yearly.csv", header = TRUE)

line1_1 = data_frame(x = c(0, 2000), y = c(0, 2000))

(ggplot(data = data_yearly) + 
 geom_point(mapping = aes(x = Precip, y = TerraClimate), size = 1) +
 geom_point(mapping = aes(x = Precip, y = ERA5_Monthly_Aggregate), color = 'red', size = 1) +
 facet_wrap('~Station') +
 geom_line(data = line1_1, mapping = aes(x = x, y = y), color = 'blue', linetype='dashed'))
```
