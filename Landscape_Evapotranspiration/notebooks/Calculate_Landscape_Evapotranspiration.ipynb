            برای راحتی استفاده از ضرایب تعریف شده در بالا، همه آن ها در یک متغییر تجمیع می شوند:        
    </font>

```python
tmp1 = Ks.merge(
    right=Kd,
    how="outer",
    on=["نوع گونه", "کد نوع گونه", "ماه"]
)

tmp2 = IE.merge(
    right=Kmc,
    how="outer",
    on=["نوع لکه", "کد نوع لکه"]
)

Coeff = tmp2.merge(
    right=tmp1,
    how="outer",
    on=["نوع گونه", "کد نوع گونه"]
)

Coeff = Coeff[["نوع لکه",
               "کد نوع لکه",
               "نوع گونه",
               "کد نوع گونه",
               "روش آبیاری",
               "کد روش آبیاری",
               "ماه",
               "Ks",
               "Kd",
               "Kmc",
               "IE"]]

Coeff.columns = ["نام نوع لکه",
                 "نوع لکه",
                 "نام نوع قلم",
                 "نوع قلم",
                 "نام زیرمجموعه هر قلم",
                 "زیرمجموعه هر قلم",
                 "ماه",
                 "Ks",
                 "Kd",
                 "Kmc",
                 "IE"]

Coeff = Coeff.pivot(
    index=["نام نوع لکه", "نوع لکه", "نام نوع قلم", "نوع قلم", "نام زیرمجموعه هر قلم", "زیرمجموعه هر قلم"],
    columns='ماه', 
    values=["Ks", "Kd", "Kmc", "IE"]
).reset_index()

Coeff.columns = ['_'.join(col).strip() for col in Coeff.columns.values]

Coeff.rename(columns={"نام نوع لکه_" : "نام نوع لکه",
                     "نوع لکه_" : "نوع لکه",
                     "نام نوع قلم_" : "نام نوع قلم",
                     "نوع قلم_" : "نوع قلم",
                     "نام زیرمجموعه هر قلم_" : "نام زیرمجموعه هر قلم",
                     "زیرمجموعه هر قلم_" : "زیرمجموعه هر قلم"},
             inplace=True)
```

```python
IDs_Asle = ["08", "09"]

dataset_coeff = pd.DataFrame()

for idf, df in dataset.groupby(["منطقه", "ناحیه", "پیمان", "آدرس"]):
    
    if df.query('`زیرمجموعه هر قلم` != @IDs_Asle').shape[0] != 0:        
    
        tmp = df.query('`زیرمجموعه هر قلم` != @IDs_Asle')\
        .groupby(["منطقه", "ناحیه", "پیمان", "آدرس", "نوع قلم"])\
        .apply(lambda x : x['مقدار آیتم'].sum())\
        .reset_index()\
        .rename(columns={0 : "مساحت گونه"})\
        .assign(**{"درصد مساحت گونه در لکه" : lambda x : x['مساحت گونه'] / x['مساحت گونه'].sum()})

        df = df.merge(tmp, how='left', on=["منطقه", "ناحیه", "پیمان", "آدرس", "نوع قلم"])

        df.loc[df["زیرمجموعه هر قلم"] == "08", ["مساحت گونه", "درصد مساحت گونه در لکه"]] = 0
        df.loc[df["زیرمجموعه هر قلم"] == "09", ["مساحت گونه", "درصد مساحت گونه در لکه"]] = 0

        Number_Species = tmp["نوع قلم"].unique().__len__()

        if "08" in list(df["زیرمجموعه هر قلم"]):
            Number_Species += 1

        df["تعداد گونه در لکه"] = Number_Species

        df['Kd_v2'] = list(map(Kd_v2.get, df['تعداد گونه در لکه']))

        df = df.merge(Coeff.drop(["نام نوع لکه", "نام نوع قلم", "نام زیرمجموعه هر قلم"], axis=1),
                      how="left",
                      on=["نوع لکه", "نوع قلم", "زیرمجموعه هر قلم"])
        
        
#         for index, row in df.iterrows():
#             if row['نوع قلم'] == "07":
#                 if row['زیرمجموعه هر قلم'] != "08" and row['زیرمجموعه هر قلم'] != "09":
#                     tree_Area = row['مقدار آیتم'] / row['مقدار آیتم خام']
#                     if tree_Area == 9:
#                         df.loc[index, 'Kd_M01'] = 0.52
#                         df.loc[index, 'Kd_M02'] = 0.52
#                         df.loc[index, 'Kd_M03'] = 0.52
#                         df.loc[index, 'Kd_M04'] = 0.52
#                         df.loc[index, 'Kd_M05'] = 0.52
#                         df.loc[index, 'Kd_M06'] = 0.52
#                         df.loc[index, 'Kd_M07'] = 0.52
#                         df.loc[index, 'Kd_M08'] = 0.52
#                         df.loc[index, 'Kd_M09'] = 0.52
#                         df.loc[index, 'Kd_M10'] = 0.52
#                         df.loc[index, 'Kd_M11'] = 0.52
#                         df.loc[index, 'Kd_M12'] = 0.52
#                     else:
#                         df.loc[index, 'Kd_M01'] = 0.65
#                         df.loc[index, 'Kd_M02'] = 0.65
#                         df.loc[index, 'Kd_M03'] = 0.65
#                         df.loc[index, 'Kd_M04'] = 0.65
#                         df.loc[index, 'Kd_M05'] = 0.65
#                         df.loc[index, 'Kd_M06'] = 0.65
#                         df.loc[index, 'Kd_M07'] = 0.65
#                         df.loc[index, 'Kd_M08'] = 0.65
#                         df.loc[index, 'Kd_M09'] = 0.65
#                         df.loc[index, 'Kd_M10'] = 0.65
#                         df.loc[index, 'Kd_M11'] = 0.65
#                         df.loc[index, 'Kd_M12'] = 0.65

        dataset_coeff = pd.concat([dataset_coeff, df])
        
    else:
        df["مساحت گونه"] = df["مقدار آیتم"] * 9
        df = df.assign(**{"درصد مساحت گونه در لکه" : 1})
        df = df.assign(**{"تعداد گونه در لکه" : 1})
        df = df.assign(**{"Kd_v2" : list(map(Kd_v2.get, df['تعداد گونه در لکه']))})
        
        df = df.merge(Coeff.drop(["نام نوع لکه", "نام نوع قلم", "نام زیرمجموعه هر قلم"], axis=1),
                      how="left",
                      on=["نوع لکه", "نوع قلم", "زیرمجموعه هر قلم"])
        
        dataset_coeff = pd.concat([dataset_coeff, df])

dataset_coeff = dataset_coeff.reset_index(drop=True)
```

# Calculate Hydromodole for Existed Layer

```python
IDs_Asle = ["08", "09"]

df = result_method_01.query('`زیرمجموعه هر قلم` != @IDs_Asle').reset_index(drop=True)

df = df[[
    "منطقه",
    "ناحیه",
    "پیمان",
    "آدرس",
    "نام نوع لکه",
    "مقدار آیتم",
    "IR",
    "IR_M04"
]] 
    
df.columns = [
    "منطقه",
    "ناحیه",
    "پیمان",
    "آدرس",
    "نوع لکه",
    "مساحت",
    "نیاز آبی سالانه",
    "نیاز آبی تیر ماه"
]

Area = df.pivot_table(
    values=[
        "مساحت"
    ],
    index=[
        "منطقه",
        "ناحیه",
        "پیمان",
        "آدرس",
        "نوع لکه"
    ],
    aggfunc={
        "مساحت": "sum"
    }
).fillna(0).reset_index(drop=False)

IR = df.pivot_table(
    values=[
        "نیاز آبی سالانه",
        "نیاز آبی تیر ماه"
    ],
    index=[
        "منطقه",
        "ناحیه",
        "پیمان",
        "آدرس",
        "نوع لکه"
    ],
    aggfunc={
        "نیاز آبی سالانه": "sum",
        "نیاز آبی تیر ماه" : "sum"
    }
).fillna(0).reset_index(drop=False)

Output_Exsited = pd.merge(
    left=Area,
    right=IR,
    how='outer', 
    on=[
        "منطقه",
        "ناحیه",
        "پیمان",
        "آدرس",
        "نوع لکه"
    ]
)

Output_Exsited_Layer = Output_Exsited


# Extract Hydromodol for Region - District

Output_Exsited = Output_Exsited.groupby(['منطقه', 'ناحیه', 'نوع لکه']).agg({
    'مساحت': 'sum',
    'نیاز آبی سالانه': 'sum',
    'نیاز آبی تیر ماه': 'sum'
}).rename(columns={
    'مساحت' : 'Area_m2',
    'نیاز آبی سالانه' : 'IWR_Year_m3',
    'نیاز آبی تیر ماه' : 'IWR_Tir_m3'
})

Output_Exsited = Output_Exsited.assign(
    Hy_Tir_m3h = (Output_Exsited['IWR_Tir_m3'] / (Output_Exsited['Area_m2'] / 10000)),
    Hy_Tir_lsh = (Output_Exsited['IWR_Tir_m3'] * 1000 / 31 / 24 / 3600) / (Output_Exsited['Area_m2'] / 10000)
).reset_index()



date = datetime.today().strftime('%Y%m%d')
file_name = PROJECT_PATH + "assets/report/Hydromodol_Existed_Layer_" + date + ".xlsx"
with pd.ExcelWriter(path=file_name, engine="openpyxl", mode="w") as writer:
    Output_Exsited.pivot(index=['منطقه', 'ناحیه'], columns='نوع لکه', values='Hy_Tir_lsh').reset_index().to_excel(writer, index=False, sheet_name="Sheet1")
```

# Calculate Hydromodole for Tafsili And Developing Layer

```python
pd.options.display.max_rows = 20

Parks = [
    'پارک های زیر 6 هکتار',
    'پارک های بالای 10 هکتار',
    'پارک های بین 6 تا 10 هکتار'
]


a = Output_Exsited_Layer.groupby(['منطقه','نوع لکه']).agg({
    'مساحت': 'sum',
    'نیاز آبی سالانه': 'sum',
    'نیاز آبی تیر ماه': 'sum'
}).reset_index().query('`نوع لکه` == @Parks')


b = a.groupby(['منطقه']).agg({
    'مساحت' : 'sum'
}).reset_index().rename(columns={'مساحت': 'مساحت کل'})


c = pd.merge(
    a,
    b,
    how='outer',
    on='منطقه'
).rename(columns={
    'منطقه' : 'Region',
    'نوع لکه' : 'Spot',
    'مساحت' : 'Area_m2',
    'نیاز آبی سالانه' : 'IWR_Year_m3',
    'نیاز آبی تیر ماه' : 'IWR_Tir_m3',
    'مساحت کل' : 'Total_Area_m2'
})

c = c.assign(
    Hy_Tir_m3h = (c['IWR_Tir_m3'] / (c['Area_m2'] / 10000)),
    Hy_Tir_lsh = (c['IWR_Tir_m3'] * 1000 / 31 / 24 / 3600) / (c['Area_m2'] / 10000)
)


c['x'] = c['Area_m2'] * c['Hy_Tir_m3h'] / c['Total_Area_m2']
c['y'] = c['Area_m2'] * c['Hy_Tir_lsh'] / c['Total_Area_m2']


c = c.groupby(['Region']).agg({
    'x' : 'sum',
    'y' : 'sum'
}).reset_index().rename(columns={
    'x' : 'Hy_Tir_m3h',
    'y' : 'Hy_Tir_lsh'
})


date = datetime.today().strftime('%Y%m%d')
file_name = PROJECT_PATH + "assets/report/Hydromodol_Tafsili_Developing_Layer_" + date + ".xlsx"
with pd.ExcelWriter(path=file_name, engine="openpyxl", mode="w") as writer:
    c.to_excel(writer, index=False, sheet_name="Sheet1")
```
