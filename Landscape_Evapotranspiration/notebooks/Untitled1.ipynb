```python
import pandas as pd

PROJECT_PATH = "C:/Users/POOYA/Dropbox/Work/HydroTech/Landscape_Evapotranspiration/"

data = pd.read_excel(
    PROJECT_PATH + "assets/report/data_cleansing_20210216.xlsx",
    engine = "openpyxl",
    dtype={
        "شناسه": str,
        "منطقه": str,
        "ناحیه": str,
        "پیمان": str,
        "آدرس": str,
        "نوع لکه": str,
        "نوع قلم": str,
        "زیرمجموعه هر قلم": str
    }
)

df = data.query('`نوع قلم` == "07"')

df["ID"] = df["منطقه"] + df["ناحیه"] + df["پیمان"] + df["آدرس"]
```

```python
import numpy as np
d1 = df.pivot_table(values=["مقدار آیتم خام"], index="ID", columns="نام زیرمجموعه هر قلم", aggfunc=np.sum, fill_value=0).reset_index(drop=False)
d1.columns = d1.columns.get_level_values(1)
d1.columns = ["ID", "آبیاری تانکری", "آبیاری تحت فشار", "آبیاری ثقلی", "آبیاری شلنگی", "اصله", "متر مربع"]
d2 = df.pivot_table(values=["مساحت آدرس"], index="ID", aggfunc=np.mean, fill_value=0).reset_index()
d3 = df[["ID", "نام نوع لکه"]].drop_duplicates(keep='last')

d4 = d1.merge(d2, how='outer', on='ID').merge(d3, how='outer', on='ID')

file_name = PROJECT_PATH + "assets/report/Tree_" + ".xlsx"
with pd.ExcelWriter(path=file_name, engine="openpyxl", mode="w") as writer:
    d4.to_excel(writer, index=False, sheet_name="Sheet1")
```
